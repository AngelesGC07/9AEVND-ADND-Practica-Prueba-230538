# Un viaje visual y revelador - DonorsChoose
## Ciencia de datos para el bien social: DonorsChoose

DonorsChoose es una organización sin fines de lucro con sede en EE. UU. que permite a las personas donar directamente a proyectos para aulas de escuelas públicas. Hasta enero de 2018, DonorsChoose.org había financiado un millón de proyectos. En el 77 % de las escuelas públicas de Estados Unidos, se ha solicitado al menos un proyecto a través de DonorsChoose.org. DonorsChoose ha compartido un conjunto de datos completo de todos los proyectos anteriores, los recursos relacionados y los metadatos de docentes, escuelas, donantes y donaciones. Estos datos se pueden utilizar para extraer información valiosa y crear una solución que permita a DonorsChoose.org desarrollar campañas de correo electrónico dirigidas a recomendar proyectos específicos para aulas a donantes anteriores. En este cuaderno, he realizado un análisis y una exploración exhaustivos del conjunto de datos compartido. Este cuaderno no solo proporciona información valiosa del conjunto de datos de DonorsChoose, sino que también sirve como tutorial para realizar análisis de datos, análisis exploratorio de datos y visualizaciones de datos.
